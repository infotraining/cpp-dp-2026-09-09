```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def draw(self):
        print(f"Drawing a circle with radius {self.radius}")

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def draw(self):
        print(f"Drawing a rectangle with width {self.width} and height {self.height}")

class IncompatibleShape:
    def render(self):
        print("Drawing an incompatible shape")
        
    def draw(self):
        self.render()

def client_renderer(shape):
    shape.draw()  # duck typing (polymorphism based on method presence)

def main():
    circle = Circle(5)
    rectangle = Rectangle(4, 6)
    incompatible_shape = IncompatibleShape()

    client_renderer(circle)
    client_renderer(rectangle)
    client_renderer(incompatible_shape)  # This would raise an AttributeError because 'IncompatibleShape' has no 'draw' method
    
main()
```
